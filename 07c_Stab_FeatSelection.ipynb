```python
def rank_features_by_fis(feature_scores, descending=True):
    """
    Rank features according to their FIS scores.

    Parameters
    ----------
    feature_scores : list of tuples
        List of (feature_name, FIS_score).
    descending : bool, optional
        If True (default), ranks from highest to lowest FIS.
        If False, ranks from lowest to highest.

    Returns
    -------
    list
        Ordered list of feature names.
    """
    sorted_features = sorted(
        feature_scores,
        key=lambda x: x[1],
        reverse=descending
    )

    return [feature for feature, _ in sorted_features]
```

```python
#

import itertools
import numpy as np
import matplotlib.pyplot as plt


sequences = [S1,S2,S3,S4,S5,S6,S7,S8,S9,S10]

# Check that all sequences contain the same features

n = len(S1)

for i,seq in enumerate(sequences):

    if len(seq) != n:
        raise ValueError(f"Sequence {i+1} has incorrect length.")

    if set(seq) != set(S1):
        raise ValueError(f"Sequence {i+1} does not contain the same features.")

# Kuncheva consistency index

def consistency_index(A, B, n):

    k = len(A)

    if k == 0 or k == n:
        return np.nan

    r = len(set(A).intersection(set(B)))

    IC = (r*n - k*k)/(k*(n-k))

    return IC

# Stability index

K = len(sequences)

stability = []

for k in range(1,n):

    pairwise = []

    for i,j in itertools.combinations(range(K),2):

        A = sequences[i][:k]
        B = sequences[j][:k]

        pairwise.append(consistency_index(A,B,n))

    stability.append(np.mean(pairwise))

# Display results

print("Subset Size    Stability")

for k,IS in enumerate(stability,start=1):
    print(f"{k:>4}          {IS:.4f}")

# Plot

plt.figure(figsize=(7,5))
plt.plot(range(1,n),stability,'s--',linewidth=1,color='red')
plt.xlabel("Number of selected features (k)", fontsize=18)
plt.ylabel("Kuncheva Stability Index $I_S$", fontsize=18)
plt.grid(True, alpha=0.3)
# plt.title("Kuncheva Stability Index")
# plt.grid(True)

plt.tight_layout()

savefig('./figs/Stab_Analy',dpi=1200,quality=100)
# plt.show()
```

```python
# Kuncheva and Jaccard Indices

import itertools
import numpy as np
import matplotlib.pyplot as plt


sequences = [S1,S2,S3,S4,S5,S6,S7,S8,S9,S10]

# Check that all sequences contain the same features

n = len(S1)

for i,seq in enumerate(sequences):

    if len(seq) != n:
        raise ValueError(f"Sequence {i+1} has incorrect length.")

    if set(seq) != set(S1):
        raise ValueError(f"Sequence {i+1} does not contain the same features.")

# Kuncheva consistency index

def consistency_index(A, B, n):

    k = len(A)

    if k == 0 or k == n:
        return np.nan

    r = len(set(A).intersection(set(B)))

    IC = (r*n - k*k)/(k*(n-k))

    return IC


# Jaccard index
def jaccard_index(A, B):

    A = set(A)
    B = set(B)

    union = len(A.union(B))

    if union == 0:
        return np.nan

    return len(A.intersection(B))/union


# Stability index

K = len(sequences)

stability = []
jaccard_stability = []

for k in range(1,n):

    pairwise = []
    pairwise_jaccard = []

    for i,j in itertools.combinations(range(K),2):

        A = sequences[i][:k]
        B = sequences[j][:k]

        pairwise.append(consistency_index(A,B,n))
        pairwise_jaccard.append(jaccard_index(A,B))

    stability.append(np.mean(pairwise))
    jaccard_stability.append(np.mean(pairwise_jaccard))


# Display results

print("Subset Size    Kuncheva    Jaccard")

for k,(IS,JS) in enumerate(zip(stability,jaccard_stability),start=1):
    print(f"{k:>4}          {IS:.4f}      {JS:.4f}")


# Plot

plt.figure(figsize=(7,5))

plt.plot(range(1,n), stability,
         's--', linewidth=1, color='red',
         label='Kuncheva')

plt.plot(range(1,n), jaccard_stability,
         'd-.', linewidth=1, color='blue',
         label='Jaccard')

plt.xlabel("Number of selected features (k)", fontsize=18)
plt.ylabel("Stability Index", fontsize=18)

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()

savefig('./figs/Stab_Analy_K_J',dpi=1200,quality=100)
# plt.show()
```
